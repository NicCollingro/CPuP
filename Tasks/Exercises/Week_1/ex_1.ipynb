# Aufgabe 1 – Die *„Basics“*
In der folgenden Code-Zelle sehen Sie, wie man Variablen erzeugt und diese anschließend ausgibt.

In Python müssen *Datentypen* nicht explizit *deklariert* werden. Anders als z.B. in *C, C++ oder Java* schreibt man den Datentyp (Zahl, Text, Liste usw.) also **nicht** vor den Variablennamen. Im folgenden Beispiel wird mit *`char_1 = "H"`* der Variable *`char_1`* der Text *`"H"`* zugewiesen. Python erkennt selbst, dass es sich um einen **String** (Zeichenkette) handelt. Einen eigenen Datentyp für einzelne Zeichen gibt es in Python übrigens nicht: `"H"` ist einfach ein String der Länge 1.

Um etwas in der *Konsole* bzw. im Jupyter-Notebook unterhalb der Zelle auszugeben, kann man unter anderem die eingebaute Funktion *`print()`* verwenden. Funktionen bekommen im Allgemeinen beliebig viele Eingaben (*Argumente*), führen damit Operationen aus und geben ggf. ein Ergebnis zurück. Erkennbar sind Funktionen an den Klammern *`()`* hinter dem *Funktionsnamen*. Funktionsname und Klammern zusammen bilden den *Funktionsaufruf*.

Im folgenden Beispiel wird zunächst der Ausdruck *`char_1 + char_2 + char_3 + char_3 + char_4 + char_5`* ausgewertet. Der Operator `+` hängt Strings aneinander (*Konkatenation*), das Ergebnis ist also `"HALLO!"`. Dieses Ergebnis wird anschließend an die Funktion *`print()`* *übergeben* und durch diese ausgegeben. Warum diese Reihenfolge wichtig ist, werden wir zu einem späteren Zeitpunkt noch genauer betrachten.

Führen Sie die Zelle mit `Shift + Enter` aus.

```python
char_1 = "H"
char_2 = "A"
char_3 = "L"
char_4 = "O"
char_5 = "!"

print(char_1 + char_2 + char_3 + char_3 + char_4 + char_5)
```

Den Datentyp *String* (eine beliebig lange Zeichenkette) haben Sie grade kennengelernt. Nun wird es Zeit, Zahlen einzuführen. Python unterscheidet dabei zwischen **ganzen Zahlen** (*integer*, kurz *`int`*, z. B. `-3`, `0`, `42`) und **Fließkommazahlen** (*floating point numbers*, kurz *`float`*, z. B. `3.14` oder `-0.5`). Beachten Sie, dass in Python ein **Punkt** statt eines Kommas verwendet wird.

Welchen Datentyp eine Zahl hat, erkennt Python automatisch. Den Unterschied sollten Sie trotzdem kennen:

- **`int`** kann in Python beliebig groß werden und wird immer **exakt** gespeichert.
- **`float`** wird intern ähnlich wie die wissenschaftliche Schreibweise gespeichert (z. B. $1{,}5 \cdot 10^3$), allerdings im Binärsystem und mit einer begrenzten Anzahl an Stellen (ca. 15–16 gültige Dezimalstellen). Viele Zahlen, etwa `0.1` oder $\frac{1}{3}$, lassen sich so nicht exakt darstellen. Dadurch kann es zu kleinen **Rundungsfehlern** kommen.

Als kleines Beispiel sehen Sie im folgenden Code zwei Fließkommazahlen *`a`* und *`b`*.

```python
a = 0.1
b = 0.2
```

Mit dem *Operator* `==` kann man prüfen, ob zwei Werte gleich sind. Das Ergebnis ist entweder *`True`* (wahr) oder *`False`* (falsch). Vergleicht man nun die Summe aus *`a`* und *`b`* mit *`0.3`*, erhält man überraschenderweise *`False`*:

```python
print(a + b == 0.3)
print(a + b)
```

Die zweite Zeile zeigt den Grund: `0.1 + 0.2` ergibt im Computer `0.30000000000000004`, also minimal mehr als `0.3`. Das ist genau der oben beschriebene Rundungsfehler.

Keine Sorge, falls das gerade sehr schnell ging: Vergleiche, `True`/`False` und den richtigen Umgang mit Rundungsfehlern schauen wir uns in den kommenden Wochen noch ausführlich an. Hier sollten Sie nur einen ersten Eindruck bekommen.

# Aufgabe 1.2 – Rechenoperationen

Im vorherigen Abschnitt haben wir bereits `int` und `float` eingeführt. Nun ist es Zeit, mit ihnen zu rechnen. Die Grundrechenarten `+`, `-`, `*` und `/` funktionieren wie gewohnt. Beachten Sie aber: Die Division `/` liefert in Python **immer** einen `float`, auch wenn das Ergebnis aufgeht (`6 / 3` ergibt `2.0`). In C oder Java würde `1 / 3` übrigens `0` ergeben!

Zusätzlich bietet Python weitere Operatoren:

| Operator | Bedeutung | Beispiel | Ergebnis |
|-|-|-|-|
| `//` | ganzzahlige Division (rundet **ab**) | `7 // 2` | `3` |
| `%` | Modulo: Rest der Division | `7 % 2` | `1` |
| `**` | Potenz | `2 ** 3` | `8` |

**Achtung:** `^` ist in Python **keine** Potenz! Für sehr große oder kleine Zahlen gibt es die wissenschaftliche Schreibweise: `3e8` steht für $3 \cdot 10^8$.

Führen Sie die folgende Zelle aus und vergleichen Sie die Ergebnisse mit Ihren Erwartungen.

```python
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2)
print(1 / 3)
print(7 // 2)
print(7 % 2)
print(2 ** 3)
print(3e8)
```
